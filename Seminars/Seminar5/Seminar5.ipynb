# 1. Введение

**Молекулярный докинг** – это широко применяемый (еще c 1980x годов) метод, позволяющий предсказать, как одна молекула (**лиганд**) связывается с другой (**рецептором**). 

Хотя изначально молекулярный докинг был разработан для предсказания взаимодействий малых молекул с белками, к настоящему времени появились методы докинга "белок-белок", "ДНК/РНК-лиганд" и даже комплексов "ДНК/РНК-белок-лиганд". 

**Задача докинга** — найти наиболее вероятную ориентацию и конформацию лиганда в активном сайте белка с учетом геометрии и энергии взаимодействия.

## Место докинга в drug design

<img src="Images/ijms-20-04331-g001.png" alt="drawing" style="width:800px;"/>

https://www.mdpi.com/1422-0067/20/18/4331

В настоящее время молекулярный докинг используется: 
* для предсказания активности лигандов по отношению к интересующей мишени
* для проведения structure-based (т.е. для известной структуры белка-рецептора) виртуального скрининга потенциальных лекарственных соединений
* для идентификации ряда мишеней, к которым лиганды демонстрируют хорошую комплементарность (фишинг и профилирование мишеней), некоторые из которых потенциально ответственны за непредвиденные побочные реакции на лекарственные препараты (прогнозирование нецелевых реакций)
* для идентификации лигандов, которые одновременно связываются с пулом интересующих мишеней (полифармакология)
* для выявления новых областей применения химических соединений с уже оптимизированными профилями безопасности (репозиционирование лекарств)

> Важно: сам по себе докинг не дает надежного результата. Его необходимо подтверждать экспериментальными данными и расчетами молекулярной динамики (МД).

# 2. Процедура докинга

<img src="Images/ijms-11-03016f1.png" alt="drawing" style="width:800px;"/>

https://www.mdpi.com/1422-0067/11/8/3016

Процедура докинга состоит из двух основных этапов:

* **Sampling** — генерация всевозможных ориентаций и конформаций лиганда (т.н. **позы лиганда**) в активном сайте белка
* **Scoring** — ранжирование полученных поз для определения наиболее вероятной (энергетически выгодной)

## 2.1. Типы докинга

Процедуры докинга можно разделить на 2 основных класса:

* **Жесткий (rigid) докинг** рассматривает лиганд и рецептор как два жестких тела, исследуя только шесть степеней свободы (перемещения и вращения). Большинство программ для докинга начинают именно с такого подхода.

* **Гибкий (flexible) докинг**, в простейшем случае, рассматривает гибкий лиганд при фиксированном жестком рецепторе, учитывая лишь конформационное пространство лиганда. 
    
  * Из-за большого размера белков и множества степеней свободы именно учет их гибкости представляет наибольшую сложность в докинге. Для учета гибкости белков используют методы: soft docking, side-chain flexibility, molecular relaxation и ensemble docking.

## 2.2. Sampling algoritms

<img src="Images/Search_alg.png" alt="drawing" style="width:400px;"/>

Основная цель алгоритма — получить всевозможные положения лиганда в активном центре белка для их дальнейшей оценки с помощью оценочной функции (scoring fucntion, см. ниже). 

**Систематические или прямые методы**

* **Конформационный поиск**
  * Последовательный полный перебор всех дихедральных, трансляционных и вращательных степеней свободы лиганда относительно белка

* **Фрагментация**
  * В этом методе лиганд разбивается на несколько фрагментов, которые поочередно "дочатся" с белком, что уменьшает вычислительную сложность:
    * либо каждый из фрагментов отдельно "дочится" в активный сайт рецептора, и затем фрагменты соединяются связями
    * Либо сначала "дочится" первый фрагмент, затем второй с учетом фиксированного первого и тд.
  * Инструменты: Flex XTM, DOCK, LUDI и др.

* **Поиск в базе данных**
  * Использование заранее подготовленных конформаций лигандов для жесткого докинга. 
  * Инструмент: FLOG


**Стохастические или случайные методы**

* **Монте-Карло**
  * Этот подход включает в себя произвольное размещение лигандов в сайте связывания рецептора, оценку его состояния и последующее создание новой конфигурации
  * Инструменты: MCDOCK, ICM и др.

* **Генетические алгоритмы**
  * Начинается с популяции поз, где конфигурация и расположение относительно рецептора описываются «геном», а оценка представляет собой «приспособленность». Выполняет преобразования, гибридизацию и т.д. наиболее приспособленных особей для создания следующего поколения и повторяет соглашение
  * Инструменты: GOLD, AutoDock и др.

* **Поиск с запретами**
  * введение ограничений для предотвращения повторного изучения уже исследованных областей конформационного пространства лигандов
  * Инструменты: PRO LEADS, Molegro Virtual Docker (MVD)TM и др.


> Дополнительно: https://www.nature.com/articles/s41598-023-40160-2

## 2.3. Scoring functions

<img src="Images/Scoring_functions.jpg" alt="drawing" style="width:1000px;"/> 

1. **Physics-based/force-field based** 

* Оценочные функции, основанные на силовых полях, представляют собой сумму двух энергий: энергии взаимодействия белок-лиганд ($E_{PL}$) и внутренней энергии лиганда ($E_{L}$), содержащих Ван-дер-Ваальсовы и электростатические термы. 
    * Терм энергии Ван-дер-Ваальса может быть задан функцией потенциала Леннарда-Джонса с различными параметрами
    * Электростатический член описывается в форме закона Кулона 
    * Влияние растворителя неявно учитывается путем введения диэлектрической проницаемости, зависящей от расстояния, в кулоновскую формулу, что уменьшает вклад взаимодействий заряд-заряд, имитируя эффект экранирования. 
    * Кроме того, СФ могут включать в себя дополнения, такие как член конфигурационной энтропии лигандов и явный учет водородных связей белок-лиганд (G-score).

* Параметры силового поля получают из экспериментальных данных и из квантово-механических расчётов *ab initio*
  
* Основные преимущества: физическая интерпретируемость, хорошая обоснованность.

* Недостатки: сложности, связанные с учетом растворителя, термодинамических эффектов и конфигурационной энтропии.



<img src="Images/SF_ff.png" alt="drawing" style="width:600px;"/> 

См. `Materials/2016-IGIGlobal-ScoringFunction.pdf`

2. **Empirical**

* Представляют собой сумму взвешенных энергетических членов, полученных в результате регрессии по экспериментальным данным.

* Учитываются вклады: водородных связей, гидрофобных и ионных взаимодействий, стерических столкновений и т.д.

* Примеры: ChemScore, X-Score, F-Score.

* Преимущества: быстрота расчетов, хорошая предсказательная способность при правильной калибровке.

* Недостатки: зависимость от обучающего набора, риск переобучения и невосприимчивость к новым типам взаимодействий.


<img src="Images/SF_emp.png" alt="drawing" style="width:600px;"/> 

3.  **Knowledge-Based**
* Используют статистические потенциалы, полученные из структурных баз данных, основанные на частоте появления атомных пар.
* Общий принцип: взаимодействия комплексов белок-лиганд моделируются с помощью парных потенциалов. Парные потенциалы получаются непосредственно из частоты встречаемости пар атомов в базе данных с помощью обратного соотношения Больцмана (см. схему ниже).

* Примеры: DrugScore, PMF, ITScore.

* Преимущества: быстрый расчет, хорошая общая применимость, возможность учета структурных особенностей.

* Недостатки: сложность определения правильной базы для формирования потенциалов.



<img src="Images/SF_kb.webp" alt="drawing" style="width:400px;"/> 

4. **Machine-Learning**

* Новые методы, использующие алгоритмы нейросетей, случайного леса и SVM для предсказания связывания.

* Обучаются на больших датасетах структур-лигандов с известными аффинностями.

* Пример: RF-Score, NNScore.

* Преимущества: высокая точность при обучении на больших объемах данных.

* Недостатки: риск переобучения, слабое предсказание при новых структурах без подходящей обучающей базы.

## 4.1. AutoDock Vina

Статья авторов метода: https://pmc.ncbi.nlm.nih.gov/articles/PMC3041641/. 

> *AutoDock Vina was developed more recently to fulfill the need for a turnkey docking method that **does not require extensive expert knowledge from users*** [ https://www.nature.com/articles/nprot.2016.051 ]

* AutoDock Vina использует сложный алгоритм поиска, включающий в себя стохастическую глобальную оптимизацию и методы локальной оптимизации на основе градиента.

* Оценочная функция Vina представляет собой гибрид эмпирической и knowledge-based функции, вдохновлённой функцией X-Score

**Термы оценочной функции AutoDockVina**

> AutoDock Vina scoring function is highly approximate, with spherically symmetric hydrogen bond potentials, implicit hydrogens and **no electrostatic contribution**. 
> 
> It has been demonstrated to perform well with ligands with typical biological size and composition [https://www.nature.com/articles/nprot.2016.051 ]


Оценочная функция AutoDock Vina по умолчанию состоит из трех стерических термов, терма водородной связи, гидрофобного термина и фактора гибкости лиганда (в зависимости от количества вращающихся связей).

## 4.3. Подготовка файлов для докинга

### 4.3.1. Подготовка белка (рецептора)

#### Получение структуры

Структура белка через protein data bank: https://www.rcsb.org/structure/7MZD.  

* Скачаем ее как `DATA/test/7MZD.pdb`

#### Подготовка файла белка для докинга 

В PyMOL удаляем все, кроме белка, и экспортируем как receptor.pdb 

В AutoDockTools: 

   1. `File -> Read Molecule -> 7MZD.pdb`
   2. Добавляем полярные водороды: `Edit -> Hydrogens -> Add -> Polar only`

<img src="Images/ADT1.png" alt="drawing" style="width:400px;"/> 

   3. Расчитываем частичные заряды : `Edit -> Charges -> Add Kolmann charges`   

> Заряды Коллмана рассчитываются по силовому полю AMBER и учитывают электронное окружение.


   4. Сохраняем как receptor.pdbqt `Grid -> Macromolecule -> Choose -> Select molecule -> save as pdbqt`

**PDBQT = (Protein Data Bank, Partial Charge (Q), & Atom Type (T))** - это формат файла, созданный на основе формата PDB и используемый в программах для молекулярного докинга, в т.ч. AutoDock. Помимо данных PDB, PDBQT содержит информацию о парциальных зарядах и типах атомов, необходимых для расчета электростатических взаимодействий. Он также описывает гибкие и жесткие части молекулы для последующего докинга.



```python
# исходный файл PDB без частичных зарядов и типов атомов
with open("DATA/receptor.pdb") as f:
    print("".join(f.readlines()[0:10]))
```

```python
#файл PDBQT с частичными зарядами и типами атомов
with open("DATA/receptor.pdbqt") as f:
    print("".join(f.readlines()[0:10]))
```

### 4.3.2. Подготовка лиганда

* Структуру лиганда можно найти в базах данных малых молекул, таких как PubChem https://pubchem.ncbi.nlm.nih.gov/ или ZINC https://cartblanche22.docking.org/.

* Для лигандов-малых молекул изначально нужна только определенная геометрия. Однако, для успешного докинга важно выбрать состояние протонирования, связи, вокруг которых возможно вращение, а также частичные заряды.

1. Найдем исходную структуру лиганда в PubChem https://pubchem.ncbi.nlm.nih.gov/
   * RTX: https://pubchem.ncbi.nlm.nih.gov/substance/363682317
   * Сохраняем ее как 3D .sdf (Если сохраняете 2D, для Vina может быть затруднительно с ним работать. Следует минимизировать структуру)
  
2. В **PyMOL** открываем `ligand.sdf`, сохраняем в формате `.mol2`.

3. В **AutoDock Tools**:

<img src="Images/ADT_lig.png" alt="drawing" style="width:500px;"/> 

  * Загружаем `ligand.mol2`
  * `Ligand → Input → Choose Ligand`
  * `Ligand → Torsion Tree → Detect root`
  * `Ligand → Torsion Tree → Choose torsions` - выбираем связи лиганда, вокруг которых разрешено вращение

  * `Ligand → Output → Save .pdbqt`
