## 1.3. Методы in silico предсказания структуры белков

<img src="Images/researchers-apply-quan-1.jpg" alt="drawing" style="width:900px;"/> 

Методы предсказания структуры белков *in silico* можно разделить на два больших класса: 
* template-based (основанные на шаблонах)
* physics-based (физически обоснованные)

> Подробно: https://biomolecula.ru/articles/torzhestvo-kompiuternykh-metodov-predskazanie-stroeniia-belkov

### 1.3.1. Template-based methods


* **Homology modeling (моделирование по гомологии)**
    * Моделирование по гомологии строит 3D структуру целевого белка на основе известной структуры похожего (гомологичного) белка — шаблона. 
    * Метод опирается на то, что эволюционно близкие белки имеют сходную пространственную структуру, даже при значительном различии последовательностей аминокислот. 
    * Сначала ищут известную структуру — шаблон, затем проводят выравнивание последовательностей целевого белка и шаблона и строят модель целевого белка на основе структуры шаблона. 
    * Метод эффективен при относительно **высокой идентичности последовательностей (>30%)**


* **Threading (Fold recognition)**
  * Метод пытается «нащупать» подходящую структурную основу, **даже если последовательности похожи слабо**. 
  * Threading сопоставляет части последовательности целевого белка с базой структур-шаблонов и выбирает те, которые лучше всего подходят по физико-химическим свойствам и структурным мотивам.


  <img src="Images/Protein-Engineering-Rational-Design-Structure-Based-Design.jpg" alt="drawing" style="width:800px;"/> 

* **Deep learning methods**
  
    Настоящий прорыв произошёл с использованием нейросетей. Эти методы не требуют жесткой зависимости от близких шаблонов и могут эффективно предсказывать структуру для широкого класса белков.

  > https://www.nature.com/articles/d41586-024-03214-7
  

* **AlphaFold (Google DeepMind)** 
  > https://habr.com/ru/articles/599703/

  * модель, которая на базе многомасштабного обучения предсказывает структуру с точностью, близкой к экспериментальным методам.
  * AlphaFold2 использует на входе следующие данные:
    * Аминокислотную последовательность целевого белка.
    * Множественное выравнивание последовательностей гомологов и информацию о близких структурных шаблонах из базы данных структур PDB.
  * Эти данные преобразуются в эмбеддинги, которые подаются в нейросеть AlphaFold2 для итеративного уточнения и построения 3D-модели белка.

* **RoseTTAFold (Institute of Protein Design, лаборатория Дэвида Бейкера)** 
  > https://biomolecula.ru/articles/konstruktorskoe-biuro-belkov

  * Rosetta рассматривает белок как конструктор LEGO, где аминокислоты — детали, взаимодействующие между собой для формирования устойчивой структуры. Задача программы — перебрать множество возможных конфигураций цепи, найти минимумы энергии, которые соответствуют стабильной, нативной структуре белка. 
  * Может работать как с «известными» шаблонами (моделирование по гомологии), так и без них (de novo), вычисляя оптимальную упаковку цепи.
  * Позволяет не только предсказывать структуру, но и улучшать свойства белков — например, увеличивать их термостабильность.

### 1.3.2. Physics-based methods

* **Ab initio modeling** 
  * Предсказание структуры основано строго на физико-химических принципах — минимизации свободной энергии, универсальных силах взаимодействия и симуляции свёртывания белка из аминокислотной последовательности.
  * Этот метод не нуждается в шаблонах и пытается вычислить наиболее энергетически выгодную конфигурацию. Подходит только для маленьких белков из-за чрезвычайной вычислительной сложности.

* **Молекулярная динамика (MD)**
  
  * MD моделирует движение белковых атомов во времени под действием физических сил, позволяя изучать их стабильность, подвижность и конформационные переходы между различными состояниями. 
  * MD часто используется для доработки моделей и исследования динамических аспектов структуры, но само по себе, чаще всего, не является методом первичного предсказания структуры белка. Хотя, есть и исключения https://biomolecula.ru/articles/millisekundnyi-barer-vziat 

* **Квантовые алгоритмы**
  * Современные разработки исследуют гибридные классическо+квантовые методы для точного моделирования белковых систем и свёртывания на уровне электронных взаимодействий, что может в будущем повысить качество предсказаний и охватить более сложные системы.

### 2.1.2. Выбор шаблонов

После завершения поиска качество шаблона оценивается двумя методами: 

1. **Глобальная оценка качества модели (GMQE)**
   * GMQE объединяет свойства, полученные из выравнивания целевой последовательности и шаблона, и структуру шаблона, и выражает ожидаемую точность или надёжность модели. 
   * GMQE принимает значения от 0 до 1, где 1 соответствует максимальной точности, а 0 — минимальной.

2. **Оценка качества четвертичной структуры (QSQE)**
   * Оценка QSQE также принимает значения от 0 до 1, однако она вычисляется только для шаблонов с наивысшим рейтингом, если существует вероятность построения олигомера. 
   * Значение выше 0,7 считается надёжными.


Кроме того, для каждого шаблона можно узнать: 

* **Экспериментальный метод**, с помощью которого была получена структура шаблона. 
  
  * Как правило, при моделировании по гомологии в качестве рекомендуется использовать рентгеновские кристаллические структуры с разрешением ниже 2,2 Å. 
  * Однако, часто приходится идти на компромисс между высокой идентичностью/сходством последовательностей и разрешением шаблона. 
  * В целом, структуры, полученные методом X-ray, предпочтительнее усреднённых структур ЯМР и Cryo-EM.
  
* **Метод поиска**, которым был найден шаблон (Blast/HHblits) 
  
* **Sequence similarity** между последовательностью и шаблоном рассчитывается с помощью матрицы замен BLOSUM62 (1 соответствует 100% сходству последовательностей, 0 - наоборот). При расчете сходства последовательностей пробелы не учитываются.

* **Лиганды** в структуре шаблона - их наличие/отсутствие может быть важно для моделирования структуры в различных состояниях связывания целевого белка с лигандами
  
* **Олигомерное состояние** - прогнозируется для каждого шаблона, но пользователь может изменить его вручную в разделе «target prediction». Если олигомерное состояние модели не полностью соответствует состоянию шаблона, то появится предупреждающий знак.
  
* **Выравнивание** последовательностей между исходной последовательностью и шаблоном

Другие вкладки: 

* **Quaternary structure**
  * отображает кластеризованные шаблоны в соответствии с их олигомерным состоянием, стехиометрией, топологией и сходством интерфейсов.
  
  * **Protein–protein interaction (PPI) Fingerprint** - это своего рода "отпечаток пальца" межсубъединичных взаимодействий (поэтому рассчитывается только для шаблонов в олигомерных состояниях). 
    * Этот показатель отражает, насколько консервативны остатки на белковых интерфейсах по сравнению с поверхностными остатками, взаимодействующими с растворителем.
      * *если специфическое взаимодействие между двумя белковыми цепями играет структурную или функциональную роль, то разумно ожидать, что остатки на соответствующем интерфейсе будут более консервативными*
  
    * Отрицательное значение консервативности интерфейса (ось y) указывает на то, что остатки интерфейса более консервативны по сравнению с поверхностными остатками.
   
    * Эта оценка консервативности получена на основе MSA гомологичных белков с использованием различных пороговых значений идентичности (ось x). 
  
    * Консервативная характеристика интерфейса может быть весьма полезна для определения того, насколько хорошо интерфейсы шаблонов адаптируются к целевому семейству белков: близкие гомологи должны достигать самых низких значений консервативности интерфейса при максимально возможном пороговом значении идентичности.

### 2.1.2. Building models

1. Чтобы построить модель из выбранного шаблона, сначала переносятся идентичные координаты атомов, моделируются вставки и неконсервативные боковые цепи аминокислот. 
   * Этот шаг выполняется движком моделирования ProMod3, который основан на фреймворке вычислительной структурной биологии OpenStructure
   * ProMod3 извлекает структурную информацию из выровненной структуры шаблона в декартовом пространстве, и, если подходящие фрагменты не находятся, используется выборка Монте-Карло для выполнения поиска в конформационном пространстве. 
  
2. Новые конформации боковых цепей, которые не могут быть найдены в шаблоне, моделируются на основе библиотеки ротамеров, зависящих от основной цепи
   
3. На заключительном этапе небольшие структурные искажения или стерические столкновения разрешаются путем короткой минимизации энергии с использованием силового поля CHARMM27 .

### 2.1.3. Оценка модели

Список построенных моделей и их характеристик отображается в разделе «Результаты модели»

Оценка качества моделей: 

* **GMQE (Global Model Quality Estimation)** - уже упомянутая выше, для предварительной оценки

* **QMEANDisCo**
  * комплексная функция оценки качества модели белка, которая дает как глобальную (для всей структуры), так и локальную (для каждого остатка) абсолютные оценки качества. 
  * Она основана на классической оценке QMEAN, дополненной прогнозом качества для каждого остатка, основанным на сопоставлении парных расстояний между остатками с констрейнтами, извлеченными из гомологичных структур. 
  * Для объединения этих данных используется нейронная сеть, обученная на предсказании локальной меры качества lDDT в диапазоне от 0 до 1. 
  * Глобальный QMEANDisCo — среднее значение по всем остаткам, с оценкой ошибки для модели данного размера.​
  * Для хорошей модели ожидается QMEANDisCo >0.70


Для каждой модели можно посмотреть на **"Structure Assessment"**. Здесь можно увидеть: 

* **Карту Рамачандрана** — двугранные углы данной структуры относительно энергетически выгодных областей двугранных углов аминокислот в целом (эти области были получены из более чем 12000 экспериментальных структур PISCES). 

* **Оценка качества структуры Molprobity**. 
  * Качество структуры выражается в индексе Molprobity, который должен быть как можно ниже, и в процентном соотношении остатков, "одобренных" Рамачандраном, в идеале выше 98%. 
  * Clash score, outliers и т.д. должны быть как можно ниже.
