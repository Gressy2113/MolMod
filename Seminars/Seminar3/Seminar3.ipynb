# 3. Выравнивание последовательностей

**Выравнивание последовательностей** — биоинформатический метод, основанный на размещении двух или более последовательностей мономеров ДНК, РНК или белков друг под другом таким образом, чтобы легко увидеть сходные участки в этих последовательностях. 

Зачем сравнивать последовательности (аминокислотные и нуклеотидные):
  * Найти похожие последовательности в базе данных
  * Для поиска мутаций в популяции или наборе генов
  * Найти консервативные для разных организмов участки, ключевые для активности белка
  * Определить вариабельные участки, которые могут объяснять различия функций белков
  * Проследить изменения, обусловленные эволюцией, которые влияют на свойства белков

## 3.1 Scoring functions

#### Простейшая функция оценки

+1 за match; 
-1 for за mismatch

````text
Seq1:  A T G C A T G
Seq2:  A T C C A T G
       | | x | | | |   →  score = 6 − 1 = 5
````

#### Более актуалная функция оценки

$$\text{Score} = \sum_{\text{pos}} \text{SIM}(s_1[\text{pos}],\, s_2[\text{pos}]) \;+\; N_{\text{gap-openings}} \cdot \text{GOP} \;+\; N_{\text{gap-positions}} \cdot \text{GEP}$$

<span style="font-size: 85%">
где:

SIM(s₁[pos], s₂[pos]) — вклад от сравнения символов в позиции pos двух выровненных последовательностей:

$$\text{SIM} = \begin{cases} +1, & \text{если совпадение (match)} \\ -1, & \text{если несовпадение (mismatch)} \end{cases}$$
Сумма $\sum_{\text{pos}}$ берётся по всем колонкам выравнивания, где нет гэпа.
GOP (Gap Opening Penalty) — штраф за открытие нового гэпа. Платится один раз в начале каждой непрерывной серии пропусков. $N_{\text{gap-openings}}$ — число таких серий в выравнивании.
GEP (Gap Extension Penalty) — штраф за продление гэпа. Платится за каждую позицию внутри гэпа. $N_{\text{gap-positions}}$ — суммарная длина всех гэпов (количество символов -).

Обычно $|\text{GOP}| \gg |\text{GEP}|$ (например, GOP = −10, GEP = −1).

<img src="Images/All_transitions_and_transversions.svg" alt="drawing" style="width:400px;"/> 

<span style="font-size: 85%">
Вероятность мутации Пиримидин->Пиримидин или Пурин-Пурин (транзиции) выше, чем Пурин-Пиримидин.
Поэтому транзиции, как правило, штрафуются слабее, чем трансляции.

### Матрицы замен аминокислот

<img src="Images/amino_prop.png" alt="drawing" style="width:400px;"/> 

* **PAM (Point Accepted Mutations)**

<img src="Images/PAM_metrix.png" alt="drawing" style="width:400px;"/> 

  * Частоты замен, полученные из выравниваний близкородственных белков (≥ 85% сходства). Химически похожие аминокислоты заменяются чаще и штрафуются слабее.
  * PAM1 ≈ в среднем ожидается одна замена 1 замена на 100 позиций.
  * Типичные варианты: PAM30 (для близких гомологов), PAM70 (средних), PAM250 (для далеких).

Сегодня PAM используют редко. Используется BLOSUM
* **BLOSUM (BLOcks SUbstitution Matrices)**
  * Проанализировали консервативные области последовательностей родственных белков, полученные из базы данных BLOCKS.
  * Разделили белки на группы по мере схожести:
    * Например, BLOSUM62 - группы состояли из белков, для которых уровень идентичности последовательностей ≥ 62%. В статистику замен идут более разнородные последовательности.
    * BLOSUM90 ≥ 90%. В статистику попадают преимущественно близкие последовательности.
  * Расчитали среднюю вероятность замен внутри каждой группы, преобразовали в веса матрицы 

Более подробно: https://bioinformaticshome.com/bioinformatics_tutorials/sequence_alignment/substitution_matrices.html

## 3.2. Типы выравниваний


* **Попарное**: 
  
  * **Глобальное**- сравнивает две последовательности по всей длине. Оно лучше всего подходит для очень похожих последовательностей примерно одинаковой длины.
  
  * **Локальное** - позволяет найти наиболее схожие участки в двух последовательностях. Может быть получено более одного выравнивания. Этот метод лучше всего подходит для последовательностей с некоторой степенью сходства или разной длины. Наиболее подходит для поиска консервативных элементов.
  
* **Множественное** (MSA - multiple sequence alignment)

### 1. Методы расстояний


Общий принцип заключается в попарном сравнении выровненных последовательностей и построении матрицы расстояний, которая затем используется для построения филогенетического дерева. 

Эти методы простые и быстрые, однако имеют недостаток - сводят данные о сходстве/различиях в последовательностях к одному числу (расстоянию), что приводит к потере информации. 

* UPGMA - невзвешенного попарного среднего
* **NJ - присоединения ближайшего соседа**
* MS - наименьших квадратов
* ME - минимальной эволюции

#### Расстояние между аминокислотными последовательностями

* **Нескорректированное расстояние (percent identity)**:
  * Простые расчеты, в которых подсчитывается количество позиций аминокислот, которые различаются в двух последовательностях
  * Расстояние часто выражается в процентах идентичности

* **Методы матрицы замен**:
  * Используют матрицы оценок (например, PAM или BLOSUM), которые отражают вероятность замены одной аминокислоты другой на основе эволюционных факторов
  * Такие подходы позволяют учитывать множественные мутации, происходящие в одной и той же позиции, и тот факт, что некоторые замены аминокислот более вероятны, чем другие

> Более подробно: https://en.wikipedia.org/wiki/Substitution_model#Amino_acid_substitution_models

#### UPGMA (Unweighted Pair Group Method with Arithmetic Mean)

* Самый простой метод
* Необходимым условием использования метода является постоянная скорость эволюции исследуемых последовательностей (гипотеза **"молекулярных часов"**). При неравномерной скорости эволюции метод может приводить к ошибкам в топологии дерева
* Генерирует укоренённые деревья 



**Алгоритм** (см. схему ниже):

>1. Находим два самых близких значения в матрице ($B$ и $F$)
>2. Объединяем соответствующие узлы в группу ($BF$)
>3. Пересчитываем расстояния до остальных узлов как среднее расстояние до первого и второго членов группы 
>$$D(X, BF) = \frac{1}{2} \cdot [D(X, B) + D(X, F)] $$
>4. Строим новую матрицу (-1 строка, -1 столбец)
>5. Если матрица все еще больше 1x1, см. п. 1

<img src="Images/UPGMA.png" alt="drawing" style="width:600px;"/>


#### NJ (Neighbor-Joining)

* Метод основывается на предположении, что наиболее вероятным будет дерево с наименьшим количеством эволюционных событий
* Генерирует неукорененные деревья

**Алгоритм**: 

>1. Начинаем со "звездной" конфигурации дерева
>2. Находим ближайшую пару узлов:

* объединяем в узел x два случайных узла i и j, все остальные вершины - другой узел
* определяем расстояние между получившимися двумя узлами

$$ S_{ij} = \frac{1}{2(n-2)} \sum_{1\leq k \leq n, k \neq i, j}  {[D(i, k) + D(j, k)]} + \frac{1}{2} D_{ij} + \frac{1}{n-2} \sum_{1\leq k \leq l \leq  n, k,l \neq i, j} {D_{kl}}   
$$

 * повторяем такой анализ для всех возможных пар узлов
 * выбираем конфигурацию, в которой это расстояние наименьшее

<img src="Images/NJ0.png" alt="drawing" style="width:600px;"/>


> 3. Пересчитываем новую матрицу (-1 строка, -1 столбец)  
>   $$d(X, AB) = \frac{1}{2} [D(X, A) + D(X, B) - D(A, B)]$$
> 4. Если матрица все еще больше 1x1, см. п. 2

<img src="Images/NJ.jpg" alt="drawing" style="width:600px;"/>


> Более подробно: https://ru.wikipedia.org/wiki/%D0%9C%D0%B5%D1%82%D0%BE%D0%B4_%D0%BF%D1%80%D0%B8%D1%81%D0%BE%D0%B5%D0%B4%D0%B8%D0%BD%D0%B5%D0%BD%D0%B8%D1%8F_%D1%81%D0%BE%D1%81%D0%B5%D0%B4%D0%B5%D0%B9
